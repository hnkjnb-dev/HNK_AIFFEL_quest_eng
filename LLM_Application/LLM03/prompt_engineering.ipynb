이번 프로젝트는 우리가 일상적으로 사용하는 유튜브, 구글과 같은 플랫폼의 알고리즘이 오히려 우리의 시야를 좁히고 있다는 문제의식에서 출발했다. 사용자는 다양한 정보를 탐색한다고 생각하지만, 실제로는 자신의 관심사와 언어에 맞춰진 콘텐츠만 반복적으로 소비하게 되는 ‘필터 버블’ 안에 머무르게 된다.

이러한 문제를 해결하기 위해, 본 프로젝트에서는 사용자가 입력한 관심 주제를 기반으로 한국 콘텐츠뿐만 아니라 해외 콘텐츠까지 함께 탐색하고 분석해주는 AI 시스템을 구현하였다. 단순히 정보를 검색하거나 번역하는 것을 넘어, 서로 다른 문화권에서 생성된 콘텐츠를 비교하고 그 차이를 분석함으로써 사용자의 시야를 확장하는 것을 목표로 한다.

특히, 사용자의 입력을 영어 검색어로 자동 변환한 뒤 글로벌 유튜브 콘텐츠를 탐색하고, 해당 콘텐츠의 텍스트뿐만 아니라 썸네일 이미지까지 함께 분석하는 멀티모달 방식을 적용하였다. 또한, 검색 결과가 부족할 경우 재검색을 수행하는 간단한 에이전트 구조를 통해 보다 적절한 정보를 수집하도록 설계하였다.

최종적으로는 한국과 해외 콘텐츠를 비교 분석하여 핵심 요약, 관점 차이, 그리고 시야 확장 인사이트를 제공함으로써, 사용자가 알고리즘에 의해 제한된 정보가 아닌 보다 넓은 관점에서 세상을 이해할 수 있도록 돕는다.

이 프로젝트는 단순한 번역이나 요약 시스템이 아니라, ‘정보의 차이를 이해하고 시야를 넓히는 경험’을 제공하는 것을 핵심 가치로 한다.

## 🚀 System Implementation Ⅰ  
### AI Startup Exploration

```python
# ================================
# 1. 라이브러리 설치
# ================================
!pip install openai youtube-search-python pillow requests httpx==0.24.1

# ================================
# 2. import
# ================================
from youtubesearchpython import VideosSearch
from openai import OpenAI
import requests
from PIL import Image
from io import BytesIO
from IPython.display import display, HTML

# ================================
# 3. API KEY 입력
# ================================
OPENAI_API_KEY = "sk-"

client = OpenAI(api_key=OPENAI_API_KEY)

# ================================
# 4. 사용자 입력
# ================================
query = input("🔍 관심 주제 입력: ")

# ================================
# 5. 한국어 → 영어 변환
# ================================
def translate(text):
    res = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "You are a translator."},
            {"role": "user", "content": f"Convert this into natural English search keywords: {text}"}
        ]
    )
    return res.choices[0].message.content.strip()

eng_query = translate(query)
print("\n🌍 영어 검색어:", eng_query)

# ================================
# 6. 유튜브 검색 함수
# ================================
def search_youtube(q):
    return VideosSearch(q, limit=3).result()

# ================================
# 7. 영상 정보 추출
# ================================
def extract(results):
    videos = []
    for v in results['result']:
        videos.append({
            "title": v['title'],
            "link": v['link'],
            "thumbnail": v['thumbnails'][0]['url'],
            "desc": v.get('descriptionSnippet', [{"text": ""}])[0]['text']
        })
    return videos

# ================================
# 8. 언어 판별 (간단)
# ================================
def detect_language(text):
    for ch in text:
        if '가' <= ch <= '힣':
            return "🇰🇷 한국어"
    return "🌍 영어"

# ================================
# 9. 썸네일 표시 + 분석 (멀티모달)
# ================================
def analyze_image(url):
    try:
        img_data = requests.get(url, timeout=5).content
        img = Image.open(BytesIO(img_data))
        display(img)

        res = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[
                {
                    "role": "user",
                    "content": [
                        {"type": "text", "text": "이 유튜브 썸네일이 어떤 내용인지 한 줄로 설명해줘"},
                        {"type": "image_url", "image_url": {"url": url}}
                    ]
                }
            ]
        )
        return res.choices[0].message.content.strip()
    except:
        return "이미지 분석 실패"

# ================================
# 10. Agent 느낌 (재검색 포함)
# ================================
def agent_search(query, eng_query):
    kr = extract(search_youtube(query))
    gl = extract(search_youtube(eng_query))

    # 글로벌 결과 부족 시 재검색
    if len(gl) < 2:
        print("\n🔁 글로벌 결과 부족 → 재검색 수행")
        gl = extract(search_youtube(eng_query + " trends"))

    return kr, gl

kr_videos, global_videos = agent_search(query, eng_query)

# ================================
# 11. 영상 출력 (이미지 + 링크 + 언어)
# ================================
print("\n==============================")
print("📺 한국 콘텐츠")
print("==============================")

kr_img_desc = []

for v in kr_videos:
    print("\n📌 제목:", v["title"])
    print("🌐 언어:", detect_language(v["title"]))
    print("🔗 링크:", v["link"])
    display(HTML(f'<a href="{v["link"]}" target="_blank">👉 영상 보러가기</a>'))

    desc = analyze_image(v["thumbnail"])
    print("🖼️ 썸네일 해석:", desc)

    kr_img_desc.append(desc)

print("\n==============================")
print("🌍 해외 콘텐츠")
print("==============================")

gl_img_desc = []

for v in global_videos:
    print("\n📌 제목:", v["title"])
    print("🌐 언어:", detect_language(v["title"]))
    print("🔗 링크:", v["link"])
    display(HTML(f'<a href="{v["link"]}" target="_blank">👉 영상 보러가기</a>'))

    desc = analyze_image(v["thumbnail"])
    print("🖼️ 썸네일 해석:", desc)

    gl_img_desc.append(desc)

# ================================
# 12. 최종 분석
# ================================
def final_analysis(topic, kr_videos, global_videos, kr_img_desc, gl_img_desc):

    prompt = f"""
주제: {topic}

[한국 콘텐츠]
{kr_videos}
이미지 해석: {kr_img_desc}

[해외 콘텐츠]
{global_videos}
이미지 해석: {gl_img_desc}

다음을 한국어로 작성:

1. 해외 콘텐츠 핵심 요약 (3줄)
2. 한국 vs 해외 관점 차이
3. 시야 확장 인사이트
"""

    res = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "You are a smart analyst."},
            {"role": "user", "content": prompt}
        ]
    )

    return res.choices[0].message.content.strip()

result = final_analysis(query, kr_videos, global_videos, kr_img_desc, gl_img_desc)

# ================================
# 13. 최종 출력
# ================================
print("\n==============================")
print("🌍 글로벌 시야 확장 결과")
print("==============================\n")

print(result)
```
